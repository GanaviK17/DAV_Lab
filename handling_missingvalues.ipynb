```python
#Median Imputation
#Median is another method for filling missing numerical values.
# Recreate the original data
data = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran', 'Meena', 'Arjun'],
    'Age': [21, 22, None, 23, 21, None],
    'Salary': [25000, None, 30000, 28000, None, 32000],
    'Department': ['CSE', 'ECE', 'CSE', None, 'ISE', 'ECE'],
    'City': ['Mysore', 'Bangalore', 'Mysore', 'Hubli', 'Bangalore', 'Mysore']
}
df1 = pd.DataFrame(data)
df1['Salary'] = df1['Salary'].fillna(df1['Salary'].median())
print(df1)
#Median is useful when the data contains outliers.
```

```python
#Forward Fill
#Forward fill (ffill) replaces a missing value with the previous available value.
print("Before Forward Fill:")
print(df1)

df1['Department'] = df1['Department'].ffill()
print("\nAfter Forward Fill:")
print(df1)
```

```python
#Machine learning algorithms generally require numerical input
#There are several encoding techniques. Label Encoding and One-Hot Encoding.
from sklearn.preprocessing import LabelEncoder
data1 = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran'],
    'Department': ['CSE', 'ECE', 'ISE', 'CSE']
}
df2 = pd.DataFrame(data1)
encoder = LabelEncoder()
df2['Department_Encoded'] = encoder.fit_transform(df2['Department'])
print(df2)
print(encoder.classes_)

```

```python
#One-hot encoding creates a separate column for each category.
data2 = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran'],
    'Department': ['CSE', 'ECE', 'ISE', 'CSE']
}
df3 = pd.DataFrame(data2)
encoded_df = pd.get_dummies(df3, columns=['Department'])
print(encoded_df)
```

```python
#If you want 0 and 1 instead of True and False:
encoded_df1 = pd.get_dummies(
    df3,
    columns=['Department'],
    dtype=int
)
print(encoded_df1)
```
